**Objective.** Predict whether a breast mass is benign or malignant using measurements of cell nuclei from fine-needle aspirate images. Malignant is the positive class (`1`); benign is `0`.

This notebook follows the Module 7 workflow: data quality and exploration, a baseline classifier, feature selection, hyperparameter tuning, and an alternative classifier. Explanations and the brief report are included in the text cells.

```python
from pathlib import Path
import sys, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from IPython.display import display
from sklearn.base import clone
from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold, cross_val_score, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve

SEED = 42
sns.set_theme(style='whitegrid', context='notebook')
pd.set_option('display.max_columns', None)
print('Python:', sys.version.split()[0], '| scikit-learn:', sklearn.__version__)
```

## 1. Import and check the data
The local UCI file has no header: the first two columns are the identifier and diagnosis, followed by 30 numerical predictors. The ten measurements are repeated as their mean, standard error (`se`), and worst value; “worst” denotes the mean of the three largest values [1]. The ID is excluded because it identifies a record rather than a measured biological characteristic.

```python
measurements = ['radius', 'texture', 'perimeter', 'area', 'smoothness',
                'compactness', 'concavity', 'concave_points', 'symmetry', 'fractal_dimension']
features = [f'{name}_{summary}' for summary in ['mean', 'se', 'worst'] for name in measurements]
data_path = Path('wdbc.data')
if not data_path.exists():
    raise FileNotFoundError('Place wdbc.data in the same working folder as this notebook.')
df = pd.read_csv(data_path, header=None, names=['id', 'diagnosis'] + features)
display(df.head())
quality = pd.Series({
    'Rows': len(df), 'Columns including ID and diagnosis': df.shape[1],
    'Predictors': len(features), 'Missing values': int(df.isna().sum().sum()),
    'Duplicate rows': int(df.duplicated().sum()),
    'Duplicate IDs': int(df.id.duplicated().sum()),
    'Duplicate predictor records': int(df[features].duplicated().sum()),
    'Infinite predictor values': int(np.isinf(df[features].to_numpy()).sum()),
    'Constant predictors': int((df[features].nunique() <= 1).sum()),
    'Negative predictor values': int((df[features] < 0).sum().sum())
}, name='Count')
display(quality.to_frame())
assert df.shape == (569, 32)
assert set(df.diagnosis) == {'B', 'M'}
assert df[features].select_dtypes(include='number').shape[1] == 30
assert df.isna().sum().sum() == 0 and np.isfinite(df[features]).all().all()
assert not df.id.duplicated().any()
display(df.diagnosis.value_counts().rename(index={'B':'Benign', 'M':'Malignant'}).to_frame('Count'))
display(df[features].describe().T.round(4))
```

The data contain 569 observations and 30 numerical predictors, with 357 benign and 212 malignant diagnoses. The checks find no missing values, infinite values, duplicate records or IDs, negative measurements, or constant predictors, so no imputation or row removal is needed. Large observations are retained because an extreme measurement is not by itself evidence of an error, and removing it could remove meaningful diagnostic information.

## 2. Training/test split and exploratory analysis
An 80/20 stratified split reserves 114 observations for testing and 455 for training, while maintaining similar class proportions. The seed is fixed before fitting any model. To prevent the exploratory analysis from guiding decisions using test information, the following plots use **training data only**; the initial quality checks simply verify file integrity.

```python
X = df[features].copy()
y = df.diagnosis.map({'B': 0, 'M': 1})
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED)
assert set(X_train.index).isdisjoint(X_test.index)
split_table = pd.DataFrame({
    'Training': y_train.value_counts().sort_index(),
    'Test': y_test.value_counts().sort_index()
}).rename(index={0:'Benign', 1:'Malignant'})
display(split_table)
eda = X_train.copy()
eda['Diagnosis'] = y_train.map({0:'Benign', 1:'Malignant'})
palette = {'Benign':'#268b8b', 'Malignant':'#cc6151'}
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
counts = eda.Diagnosis.value_counts().reindex(['Benign','Malignant'])
ax.bar(counts.index, counts.values, color=[palette[c] for c in counts.index])
for i, count in enumerate(counts):
    ax.text(i, count + 4, f'{count} ({count / len(eda):.1%})', ha='center')
ax.set(title='Diagnosis distribution - training set', ylabel='Number of observations', ylim=(0, 330))
plt.tight_layout(); plt.show()
```

**Plot rationale and interpretation.** A bar chart makes the class counts and proportions easy to compare. The training set contains 285 benign and 170 malignant observations, so the imbalance is moderate; a classifier predicting only benign would achieve about 63% accuracy but detect no malignant cases. Therefore, accuracy should be interpreted together with malignant-class precision, recall, and F1.

```python
plot_features = ['radius_mean', 'texture_mean', 'area_mean', 'concavity_mean']
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for ax, feature in zip(axes.flat, plot_features):
    sns.histplot(data=eda, x=feature, hue='Diagnosis', hue_order=['Benign','Malignant'],
                 palette=palette, bins=25, stat='density', common_norm=False,
                 element='step', alpha=0.3, ax=ax)
    ax.set_xlabel(feature.replace('_', ' ').title())
fig.suptitle('Feature distributions by diagnosis - training set')
plt.tight_layout(); plt.show()
```

**Plot rationale and interpretation.** Histograms compare the distribution shapes and overlap between diagnoses, with each class normalized separately so unequal class sizes do not dominate. Malignant observations generally have larger mean radius, area, and concavity, while mean texture shows substantial overlap. The overlap indicates that a single measurement cannot perfectly separate the two classes.

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, feature in zip(axes, ['radius_worst', 'texture_worst', 'concave_points_worst']):
    sns.boxplot(data=eda, x='Diagnosis', y=feature, order=['Benign','Malignant'],
                hue='Diagnosis', palette=palette, dodge=False, ax=ax)
    if ax.legend_ is not None: ax.legend_.remove()
    ax.set(xlabel='', ylabel=feature.replace('_', ' ').title())
fig.suptitle('Median, spread, and extreme measurements - training set')
plt.tight_layout(); plt.show()
```

**Plot rationale and interpretation.** Box plots show differences in medians and spread while highlighting observations outside the whiskers. Malignant cases tend to have higher worst radius and worst concave-point values, but the class distributions still overlap. Points beyond the whiskers are retained because the plot alone does not establish that they are recording errors.

```python
fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(X_train.corr(), cmap='vlag', center=0, vmin=-1, vmax=1,
            square=True, ax=ax, cbar_kws={'label':'Pearson correlation'})
ax.set_title('Correlations among all 30 predictors - training set')
ax.tick_params(axis='both', labelsize=8)
plt.tight_layout(); plt.show()
print('Training correlation, mean radius vs mean perimeter:',
      round(X_train.radius_mean.corr(X_train.perimeter_mean), 3))
```

**Plot rationale and interpretation.** A correlation heatmap summarizes relationships among all predictors in one view. Radius, perimeter, and area are strongly related, and several mean and worst measurements also share similar information. This motivates testing a smaller feature set, although ranking features individually does not guarantee that correlated variables will be removed.

## 3. Baseline classifier
Logistic regression is a suitable baseline because it is a simple, regularized binary classifier. Standardization is included because the predictors have different scales, and the scaler is fitted inside the pipeline using only the relevant training fold [3]. The baseline uses all 30 predictors, `C=1`, L2 regularization, and a probability threshold of 0.5.

**Evaluation plan.** Accuracy measures the fraction correctly classified; precision is the fraction of predicted malignant cases that are malignant; recall is the fraction of malignant cases detected; and F1 balances precision and recall. ROC AUC measures ranking performance over thresholds and is computed from continuous scores, not predicted class labels [4]. F1 is the preselected tuning criterion because both missed malignant cases and false alarms matter; recall and false negatives are also reported explicitly.

All stages use the same repeated stratified cross-validation (5 folds, 3 repeats) within the training set. The four configurations are fixed before their held-out test results are compared; test scores are not used to change the feature count, grid, threshold, or seed.

```python
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=1001)
baseline = Pipeline([
    ('scale', StandardScaler()),
    ('model', LogisticRegression(C=1.0, solver='liblinear', max_iter=5000, random_state=SEED))
])
baseline_cv = cross_val_score(baseline, X_train, y_train, cv=cv, scoring='f1', n_jobs=1)
baseline.fit(X_train, y_train)
print(f'Baseline training-CV F1: {baseline_cv.mean():.4f} (SD {baseline_cv.std():.4f})')
```

**Baseline performance and limitations.** On the held-out test set (computed in Section 7), the baseline achieves 97.37% accuracy, 97.56% precision, 95.24% recall, and an F1 of 0.9639, with two missed malignant cases and one false alarm. Its mean training-CV F1 is 0.9640, indicating strong performance across the predefined folds. However, a linear decision boundary may miss nonlinear patterns, and this small, single-source test set cannot establish clinical generalization.

**Proposed improvements.** Test feature selection to reduce the number of inputs, tune regularization strength and class weights, and compare a nonlinear support vector machine. These changes may help, but improvement must be measured rather than assumed.

## 4. Feature selection
`SelectKBest(f_classif, k=15)` keeps the 15 features with the highest ANOVA F-scores, measuring their individual association with diagnosis. The number 15 is fixed in advance to test a model using half the predictors; it is not selected by trying different values on the test set. Selection is inside the pipeline and is refitted within every cross-validation fold [3, 5].

```python
selected = Pipeline([
    ('select', SelectKBest(score_func=f_classif, k=15)),
    ('scale', StandardScaler()),
    ('model', LogisticRegression(C=1.0, solver='liblinear', max_iter=5000, random_state=SEED))
])
selected_cv = cross_val_score(selected, X_train, y_train, cv=cv, scoring='f1', n_jobs=1)
selected.fit(X_train, y_train)
selection_table = pd.DataFrame({
    'Feature': features,
    'Training F-score': selected.named_steps['select'].scores_,
    'Selected': selected.named_steps['select'].get_support()
}).sort_values('Training F-score', ascending=False)
display(selection_table.round(2))
print(f'Selected-feature training-CV F1: {selected_cv.mean():.4f} (SD {selected_cv.std():.4f})')
```

**Change after feature selection.** Reducing the input set from 30 to 15 features lowers mean training-CV F1 from 0.9640 to 0.9199. On the test set, accuracy remains 97.37%, but F1 decreases slightly from 0.9639 to 0.9630 and missed malignant cases increase from two to three. This reduction does not improve performance: features that are weak individually may still add useful information jointly, so the full-feature baseline remains a candidate.

## 5. Hyperparameter tuning
The selected-feature logistic regression is tuned over `C = [0.01, 0.1, 1, 10, 100]` and `class_weight = [None, 'balanced']` (10 combinations). Smaller C means stronger regularization; balanced weights increase the contribution of the less frequent malignant class. `GridSearchCV` maximizes mean training-CV F1 and refits the best configuration on the 455 training observations [6].

```python
lr_grid = {'model__C': [0.01, 0.1, 1, 10, 100],
           'model__class_weight': [None, 'balanced']}
lr_search = GridSearchCV(clone(selected), lr_grid, scoring='f1', cv=cv,
                         n_jobs=1, refit=True, error_score='raise')
lr_search.fit(X_train, y_train)
tuned = lr_search.best_estimator_
print('Best logistic regression parameters:', lr_search.best_params_)
display(pd.DataFrame(lr_search.cv_results_)[
    ['param_model__C', 'param_model__class_weight', 'mean_test_score', 'std_test_score', 'rank_test_score']
].sort_values('rank_test_score').round(4))
```

**Change after tuning.** The best selected-feature logistic regression uses `C=10` and `class_weight=None`, increasing mean training-CV F1 from 0.9199 to 0.9267. Test accuracy increases from 97.37% to 98.25%, F1 increases from 0.9630 to 0.9756, and missed malignant cases decrease from three to two, with no false alarms. Tuning partially recovers the loss in training-CV performance after feature selection, but does not exceed the full-feature baseline CV F1 of 0.9640.

## 6. Alternative classifier: nonlinear SVM
An SVM with a radial basis function (RBF) kernel allows a nonlinear decision boundary. It uses the same feature-selection rule and standardization, while a grid tests `C = [0.1, 1, 10, 100]` and `gamma = ['scale', 0.01, 0.1]` (12 combinations); gamma controls how locally each training observation influences the boundary. This compares a different model family with the same 15-feature representation.

```python
svm_pipeline = Pipeline([
    ('select', SelectKBest(f_classif, k=15)),
    ('scale', StandardScaler()),
    ('model', SVC(kernel='rbf'))
])
svm_search = GridSearchCV(svm_pipeline,
    {'model__C':[0.1, 1, 10, 100], 'model__gamma':['scale', 0.01, 0.1]},
    scoring='f1', cv=cv, n_jobs=1, refit=True, error_score='raise')
svm_search.fit(X_train, y_train)
alternative = svm_search.best_estimator_
print('Best SVM parameters:', svm_search.best_params_)
display(pd.DataFrame(svm_search.cv_results_)[
    ['param_model__C', 'param_model__gamma', 'mean_test_score', 'std_test_score', 'rank_test_score']
].sort_values('rank_test_score').round(4))
models = {'Baseline LR': baseline, 'Selected LR': selected,
          'Tuned LR': tuned, 'Tuned RBF SVM': alternative}
cv_means = [baseline_cv.mean(), selected_cv.mean(), lr_search.best_score_, svm_search.best_score_]
cv_sds = [baseline_cv.std(), selected_cv.std(),
          lr_search.cv_results_['std_test_score'][lr_search.best_index_],
          svm_search.cv_results_['std_test_score'][svm_search.best_index_]]
preferred_name = list(models)[int(np.argmax(cv_means))]
print('Preferred configuration based only on training-CV F1:', preferred_name)
```

**Change with the alternative model.** The best RBF SVM uses `C=100` and `gamma="scale"`, obtaining a mean training-CV F1 of 0.9335 compared with 0.9267 for tuned logistic regression. Its test accuracy remains 98.25%, while F1 increases slightly from 0.9756 to 0.9762 and recall increases from 95.24% to 97.62%; it misses one malignant case and produces one false alarm. The SVM performs best on test F1 among these four configurations, but the full-feature baseline remains preferred under the predefined training-CV selection rule.

## 7. Final held-out comparison
All four configurations have now been fitted without using the test set. The following comparison applies them to the same 114 held-out observations, with malignant as the positive class. Logistic regression uses its default 0.5 threshold and SVM uses its default decision-score threshold of zero; neither threshold is optimized on the test set.

```python
rows, predictions, scores = [], {}, {}
for name, model in models.items():
    pred = model.predict(X_test)
    score = model.decision_function(X_test)
    tn, fp, fn, tp = confusion_matrix(y_test, pred, labels=[0,1]).ravel()
    predictions[name], scores[name] = pred, score
    rows.append({'Model':name, 'Features':30 if name == 'Baseline LR' else 15,
        'Accuracy':accuracy_score(y_test,pred), 'Precision':precision_score(y_test,pred),
        'Recall':recall_score(y_test,pred), 'F1':f1_score(y_test,pred),
        'ROC AUC':roc_auc_score(y_test,score), 'TN':int(tn), 'FP':int(fp), 'FN':int(fn), 'TP':int(tp)})
results = pd.DataFrame(rows).set_index('Model')
results.insert(1, 'Training CV F1', cv_means)
results.insert(2, 'CV SD', cv_sds)
display(results[['Features', 'Training CV F1', 'CV SD']].round(4))
display(results[['Accuracy', 'Precision', 'Recall', 'F1', 'ROC AUC']].round(4))
display(results[['TN', 'FP', 'FN', 'TP']])
print('Training-CV selected model:', preferred_name)
print('Test-set majority-class accuracy:', (y_test == 0).mean())
print('One test error corresponds to', round(100 / len(y_test), 2), 'percentage points of accuracy.')
```

**Step-by-step interpretation.** Feature selection preserves test accuracy but exchanges one false alarm for one additional missed malignant case. Tuning the selected-feature logistic regression corrects one of those missed cases, while the SVM detects one further malignant case at the cost of one false alarm; both tuned models correctly classify 112 of 114 test observations. The SVM has the highest test recall and F1, whereas the selected and tuned logistic regressions have the highest test AUC (0.9993), illustrating that ranking performance and threshold-based performance measure different things.

Training-CV scores for the tuned models are the best scores from their searches, so they may be optimistic after selecting among configurations. The reported CV standard deviations describe variation across folds, not confidence intervals, and the repeated folds are not independent. The held-out results provide a separate check; a small difference on this one split does not establish that one algorithm is universally better.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, (name, pred) in zip(axes.flat, predictions.items()):
    sns.heatmap(confusion_matrix(y_test, pred, labels=[0,1]), annot=True, fmt='d',
                cmap='Blues', cbar=False, vmin=0, vmax=72, square=True,
                xticklabels=['Benign','Malignant'], yticklabels=['Benign','Malignant'], ax=ax)
    ax.set(title=name, xlabel='Predicted diagnosis', ylabel='Actual diagnosis')
fig.suptitle('Held-out confusion matrices: 72 benign and 42 malignant cases')
plt.tight_layout(); plt.show()
fig, ax = plt.subplots(figsize=(7, 5))
for name, score in scores.items():
    fpr, tpr, _ = roc_curve(y_test, score)
    ax.plot(fpr, tpr, label=f'{name} (AUC = {roc_auc_score(y_test, score):.4f})')
ax.plot([0,1], [0,1], '--', color='gray', label='Chance ranking')
ax.set(title='Held-out ROC curves', xlabel='False positive rate', ylabel='True positive rate'); ax.legend(loc='lower right')
plt.tight_layout(); plt.show()
```

**Plot rationale and interpretation.** Confusion matrices distinguish false alarms (upper right) from missed malignant cases (lower left), which overall accuracy can hide. ROC curves summarize discrimination over thresholds using continuous decision scores, although a high AUC does not guarantee high recall at the default threshold. The numerical comparison above should be read together with these error counts.

## 8. Conclusions and limitations
The full-feature logistic regression is a strong baseline and has the highest mean training-CV F1 (0.9640), so it remains the preferred model under the predefined selection rule. Reducing the model to 15 features lowers cross-validation performance, and subsequent tuning and a nonlinear SVM recover only part of that loss. On the held-out split, both tuned models achieve 98.25% accuracy, and the SVM detects 41 of 42 malignant cases; this small test-set advantage does not justify changing the selection rule after seeing the test results.

This is a small historical dataset from one source, and a random held-out split does not establish performance in another hospital, population, or measurement setting. The dataset contains no demographic variables for subgroup evaluation, and the model has not been externally validated or assessed for probability calibration. The ANOVA selector ranks features individually and can retain correlated predictors or discard variables useful jointly; its scores are not evidence of causal effects. Further work could use nested cross-validation for model-selection assessment, explore the feature count within training folds, and tune a recall-oriented threshold using training validation data before evaluating on an independent external dataset. These results support a classroom classification exercise, not clinical deployment.
