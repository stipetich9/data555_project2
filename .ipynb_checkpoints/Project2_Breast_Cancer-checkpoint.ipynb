The aim is to classify breast masses as benign (0) or malignant (1) using measurements of cell nuclei. The analysis checks the data, compares six models, and tests whether feature selection and tuning improve the results.

```python
from pathlib import Path
import sys, json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from IPython.display import display
from sklearn.base import clone
from sklearn.model_selection import train_test_split, RepeatedStratifiedKFold, cross_val_score, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, roc_curve

SEED = 42
sns.set_theme(style='whitegrid', context='notebook')
pd.set_option('display.max_columns', None)
print('Python:', sys.version.split()[0], '| scikit-learn:', sklearn.__version__)
```

## 1. Load and check the data
The file contains an ID, a diagnosis, and 30 numerical features. These describe ten characteristics of cell nuclei using their mean, standard error, and “worst” value, which is the average of the three largest values [1]. The ID is excluded from the models because it only identifies each record.

```python
measurements = ['radius', 'texture', 'perimeter', 'area', 'smoothness',
                'compactness', 'concavity', 'concave_points', 'symmetry', 'fractal_dimension']
features = [f'{name}_{summary}' for summary in ['mean', 'se', 'worst'] for name in measurements]
data_path = Path('wdbc.data')
if not data_path.exists():
    raise FileNotFoundError('Place wdbc.data in the same working folder as this notebook.')
df = pd.read_csv(data_path, header=None, names=['id', 'diagnosis'] + features)
display(df.head())
quality = pd.Series({
    'Rows': len(df), 'Columns including ID and diagnosis': df.shape[1],
    'Predictors': len(features), 'Missing values': int(df.isna().sum().sum()),
    'Duplicate rows': int(df.duplicated().sum()),
    'Duplicate IDs': int(df.id.duplicated().sum()),
    'Duplicate predictor records': int(df[features].duplicated().sum()),
    'Infinite predictor values': int(np.isinf(df[features].to_numpy()).sum()),
    'Constant predictors': int((df[features].nunique() <= 1).sum()),
    'Negative predictor values': int((df[features] < 0).sum().sum())
}, name='Count')
display(quality.to_frame())
assert df.shape == (569, 32)
assert set(df.diagnosis) == {'B', 'M'}
assert df[features].select_dtypes(include='number').shape[1] == 30
assert df.isna().sum().sum() == 0 and np.isfinite(df[features]).all().all()
assert not df.id.duplicated().any()
display(df.diagnosis.value_counts().rename(index={'B':'Benign', 'M':'Malignant'}).to_frame('Count'))
display(df[features].describe().T.round(4))
```

The dataset has 569 records: 357 benign and 212 malignant. There are no missing values, duplicate records or IDs, infinite values, negative measurements, or constant features, so no data need to be filled in or removed. Extreme values are kept because they may represent real differences between cases.

## 2. Split and explore the data
The data are split into 80% training (455 records) and 20% testing (114 records). A stratified split keeps similar proportions of benign and malignant cases in both groups, and a fixed random seed makes the split reproducible. The following plots use only the training data.

```python
X = df[features].copy()
y = df.diagnosis.map({'B': 0, 'M': 1})
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED)
assert set(X_train.index).isdisjoint(X_test.index)
split_table = pd.DataFrame({
    'Training': y_train.value_counts().sort_index(),
    'Test': y_test.value_counts().sort_index()
}).rename(index={0:'Benign', 1:'Malignant'})
display(split_table)
eda = X_train.copy()
eda['Diagnosis'] = y_train.map({0:'Benign', 1:'Malignant'})
palette = {'Benign':'#268b8b', 'Malignant':'#cc6151'}
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
counts = eda.Diagnosis.value_counts().reindex(['Benign','Malignant'])
ax.bar(counts.index, counts.values, color=[palette[c] for c in counts.index])
for i, count in enumerate(counts):
    ax.text(i, count + 4, f'{count} ({count / len(eda):.1%})', ha='center')
ax.set(title='Diagnosis distribution - training set', ylabel='Number of observations', ylim=(0, 330))
plt.tight_layout(); plt.show()
```

The bar chart shows how many cases belong to each class. The training set has 285 benign and 170 malignant cases, so always predicting benign would give about 63% accuracy while missing every malignant case. This is why recall and F1 are considered alongside accuracy.

```python
plot_features = ['radius_mean', 'texture_mean', 'area_mean', 'concavity_mean']
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for ax, feature in zip(axes.flat, plot_features):
    sns.histplot(data=eda, x=feature, hue='Diagnosis', hue_order=['Benign','Malignant'],
                 palette=palette, bins=25, stat='density', common_norm=False,
                 element='step', alpha=0.3, ax=ax)
    ax.set_xlabel(feature.replace('_', ' ').title())
fig.suptitle('Feature distributions by diagnosis - training set')
plt.tight_layout(); plt.show()
```

Histograms show how measurements differ between diagnoses; each class is normalized separately to make the comparison fair. Malignant cases generally have larger radius, area, and concavity values, while texture shows more overlap. The overlap suggests that several features are needed to distinguish the classes.

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, feature in zip(axes, ['radius_worst', 'texture_worst', 'concave_points_worst']):
    sns.boxplot(data=eda, x='Diagnosis', y=feature, order=['Benign','Malignant'],
                hue='Diagnosis', palette=palette, dodge=False, ax=ax)
    if ax.legend_ is not None: ax.legend_.remove()
    ax.set(xlabel='', ylabel=feature.replace('_', ' ').title())
fig.suptitle('Median, spread, and extreme measurements - training set')
plt.tight_layout(); plt.show()
```

Box plots compare the median, spread, and extreme values in each class. Malignant cases tend to have higher worst radius and worst concave-point values, although the groups overlap. Extreme points are kept because being unusual does not necessarily mean they are errors.

```python
fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(X_train.corr(), cmap='vlag', center=0, vmin=-1, vmax=1,
            square=True, ax=ax, cbar_kws={'label':'Pearson correlation'})
ax.set_title('Correlations among all 30 predictors - training set')
ax.tick_params(axis='both', labelsize=8)
plt.tight_layout(); plt.show()
print('Training correlation, mean radius vs mean perimeter:',
      round(X_train.radius_mean.corr(X_train.perimeter_mean), 3))
```

The heatmap shows relationships between the features. Radius, perimeter, and area are strongly correlated, as are several mean and worst measurements. This suggests some information is repeated and motivates testing a smaller feature set.

## 3. Compare six models
Six models were compared: logistic regression, SVM, KNN, decision tree, random forest, and Gaussian Naive Bayes. Each uses all 30 features and the same 5-fold cross-validation, repeated three times, within the training set. This means each model is trained and checked on the same smaller groups of training records.

The models are ranked by their average F1 score, which balances precision and recall. The two highest-ranked models are taken forward. Features are scaled for logistic regression, SVM, and KNN, with scaling learned only from the training portion of each fold.

These are initial models, before tuning. Their settings are shown in the code below.

```python
import os
os.environ["LOKY_MAX_CPU_COUNT"] = "1"
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_validate

cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=1001)
def scaled(model):
    return Pipeline([('scale', StandardScaler()), ('model', model)])
candidates = {
    'Logistic regression': scaled(LogisticRegression(C=1, solver='liblinear', max_iter=5000, random_state=SEED)),
    'RBF SVM': scaled(SVC(C=1, kernel='rbf', gamma='scale')),
    'KNN': scaled(KNeighborsClassifier(n_neighbors=5)),
    'Decision tree': DecisionTreeClassifier(random_state=SEED),
    'Random forest': RandomForestClassifier(n_estimators=100, random_state=SEED, n_jobs=1),
    'Gaussian Naive Bayes': GaussianNB()
}
scoring = {'Accuracy':'accuracy', 'Precision':'precision', 'Recall':'recall', 'F1':'f1', 'ROC AUC':'roc_auc'}
comparison_rows, comparison_scores = [], {}
for name, model in candidates.items():
    score = cross_validate(model, X_train, y_train, cv=cv, scoring=scoring, n_jobs=1)
    comparison_scores[name] = score
    row = {'Model':name, **{metric:score['test_' + metric].mean() for metric in scoring}}
    row['F1 SD'] = score['test_F1'].std()
    comparison_rows.append(row)
comparison = pd.DataFrame(comparison_rows).set_index('Model').sort_values('F1', ascending=False)
display(comparison.round(4))
print('Top two families based on training-CV F1:', comparison.index[:2].tolist())
comparison.to_csv('baseline_comparison.csv')
fig, ax = plt.subplots(figsize=(9, 5))
plot_order = comparison.sort_values('F1')
ax.barh(plot_order.index, plot_order['F1'], xerr=plot_order['F1 SD'], color='#268b8b', capsize=3)
ax.set(xlabel='Mean malignant-class F1 (error bars: fold SD)', title='Six baseline algorithms - training cross-validation', xlim=(0,1.06))
for i, value in enumerate(plot_order['F1']): ax.text(value + 0.045, i, f'{value:.4f}', va='center', fontsize=9)
plt.tight_layout(); plt.show()
```

The bar chart makes the F1 scores easy to compare; the error bars show how much scores vary across validation folds. Logistic regression has the highest average F1 (0.9640), followed by SVM (0.9562) and KNN (0.9536). Logistic regression is therefore used as the baseline and SVM as the alternative, although the small differences do not prove that one is always better.

## 4. Train the baseline model
Logistic regression is the baseline because it performed best in the initial comparison. It uses all 30 features, scaling, and its standard decision threshold of 0.5. The same cross-validation splits are used in the next steps, and model settings are chosen using training data only.

```python
cv = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=1001)
baseline = Pipeline([
    ('scale', StandardScaler()),
    ('model', LogisticRegression(C=1.0, solver='liblinear', max_iter=5000, random_state=SEED))
])
baseline_cv = comparison_scores['Logistic regression']['test_F1']
baseline.fit(X_train, y_train)
print(f'Baseline training-CV F1: {baseline_cv.mean():.4f} (SD {baseline_cv.std():.4f})')
```

The baseline achieves 97.37% test accuracy, 97.56% precision, 95.24% recall, and an F1 of 0.9639, missing two malignant cases and producing one false alarm (see Section 8). Its average cross-validation F1 is 0.9640. However, it may miss nonlinear patterns, and results from one small dataset may not transfer to other populations.

To try to improve the model, the next steps reduce the number of features, adjust model settings, and test SVM.

## 5. Select features
`SelectKBest` keeps the 15 features with the strongest individual relationship to diagnosis. This tests whether half the features are enough; the number 15 was not chosen using test results. Selection is repeated inside each training fold to avoid using validation information [2, 3].

```python
selected = Pipeline([
    ('select', SelectKBest(score_func=f_classif, k=15)),
    ('scale', StandardScaler()),
    ('model', LogisticRegression(C=1.0, solver='liblinear', max_iter=5000, random_state=SEED))
])
selected_cv = cross_val_score(selected, X_train, y_train, cv=cv, scoring='f1', n_jobs=1)
selected.fit(X_train, y_train)
selection_table = pd.DataFrame({
    'Feature': features,
    'Training F-score': selected.named_steps['select'].scores_,
    'Selected': selected.named_steps['select'].get_support()
}).sort_values('Training F-score', ascending=False)
display(selection_table.round(2))
print(f'Selected-feature training-CV F1: {selected_cv.mean():.4f} (SD {selected_cv.std():.4f})')
```

Using 15 features lowers average cross-validation F1 from **0.9640 to 0.9199**. Test accuracy stays at 97.37%, but F1 falls slightly to 0.9630 and the model misses three malignant cases instead of two. Feature selection does not improve this model, possibly because some discarded features are useful when combined with others.

## 6. Tune logistic regression
A grid search tests five values of `C` and two class-weight settings, giving ten combinations. `C` controls regularization: smaller values place a stronger limit on the model's coefficients, while balanced class weights give more weight to the less common class. The combination with the highest average training cross-validation F1 is selected [4].

```python
lr_grid = {'model__C': [0.01, 0.1, 1, 10, 100],
           'model__class_weight': [None, 'balanced']}
lr_search = GridSearchCV(clone(selected), lr_grid, scoring='f1', cv=cv,
                         n_jobs=1, refit=True, error_score='raise')
lr_search.fit(X_train, y_train)
tuned = lr_search.best_estimator_
print('Best logistic regression parameters:', lr_search.best_params_)
display(pd.DataFrame(lr_search.cv_results_)[
    ['param_model__C', 'param_model__class_weight', 'mean_test_score', 'std_test_score', 'rank_test_score']
].sort_values('rank_test_score').round(4))
```

The best settings are C=10 with no class weighting. Average cross-validation F1 rises from 0.9199 to 0.9267, while test accuracy rises to 98.25% and test F1 to 0.9756, with two missed malignant cases and no false alarms. Tuning improves the 15-feature model, but its cross-validation F1 remains below the 30-feature baseline.

## 7. Train an alternative model: SVM
SVM ranked second in the initial comparison and can capture nonlinear patterns using an RBF kernel. It uses the same 15-feature selection and scaling steps. A grid search tests 12 combinations of `C` and `gamma`, which control the strength and shape of the decision boundary.

```python
svm_pipeline = Pipeline([
    ('select', SelectKBest(f_classif, k=15)),
    ('scale', StandardScaler()),
    ('model', SVC(kernel='rbf'))
])
svm_search = GridSearchCV(svm_pipeline,
    {'model__C':[0.1, 1, 10, 100], 'model__gamma':['scale', 0.01, 0.1]},
    scoring='f1', cv=cv, n_jobs=1, refit=True, error_score='raise')
svm_search.fit(X_train, y_train)
alternative = svm_search.best_estimator_
print('Best SVM parameters:', svm_search.best_params_)
display(pd.DataFrame(svm_search.cv_results_)[
    ['param_model__C', 'param_model__gamma', 'mean_test_score', 'std_test_score', 'rank_test_score']
].sort_values('rank_test_score').round(4))
models = {'Baseline LR': baseline, 'Selected LR': selected,
          'Tuned LR': tuned, 'Tuned RBF SVM': alternative}
cv_means = [baseline_cv.mean(), selected_cv.mean(), lr_search.best_score_, svm_search.best_score_]
cv_sds = [baseline_cv.std(), selected_cv.std(),
          lr_search.cv_results_['std_test_score'][lr_search.best_index_],
          svm_search.cv_results_['std_test_score'][svm_search.best_index_]]
preferred_name = list(models)[int(np.argmax(cv_means))]
print('Preferred configuration based only on training-CV F1:', preferred_name)
```

The best SVM uses C=100 and gamma="scale", with an average cross-validation F1 of 0.9335. Test accuracy is 98.25% and F1 is 0.9762; it detects 41 of 42 malignant cases and gives one false alarm. It improves on the tuned 15-feature logistic regression, but its cross-validation F1 remains below both the full-feature logistic regression (0.9640) and the initial full-feature SVM (0.9562).

## 8. Compare test results
The four developed models are evaluated on the same 114 test records. Their standard decision thresholds are used, and none of their settings are chosen from these test results. ROC AUC is calculated using continuous model scores [5].

```python
rows, predictions, scores = [], {}, {}
for name, model in models.items():
    pred = model.predict(X_test)
    score = model.decision_function(X_test)
    tn, fp, fn, tp = confusion_matrix(y_test, pred, labels=[0,1]).ravel()
    predictions[name], scores[name] = pred, score
    rows.append({'Model':name, 'Features':30 if name == 'Baseline LR' else 15,
        'Accuracy':accuracy_score(y_test,pred), 'Precision':precision_score(y_test,pred),
        'Recall':recall_score(y_test,pred), 'F1':f1_score(y_test,pred),
        'ROC AUC':roc_auc_score(y_test,score), 'TN':int(tn), 'FP':int(fp), 'FN':int(fn), 'TP':int(tp)})
results = pd.DataFrame(rows).set_index('Model')
results.insert(1, 'Training CV F1', cv_means)
results.insert(2, 'CV SD', cv_sds)
display(results[['Features', 'Training CV F1', 'CV SD']].round(4))
display(results[['Accuracy', 'Precision', 'Recall', 'F1', 'ROC AUC']].round(4))
display(results[['TN', 'FP', 'FN', 'TP']])
print('Training-CV selected model:', preferred_name)
print('Test-set majority-class accuracy:', (y_test == 0).mean())
print('One test error corresponds to', round(100 / len(y_test), 2), 'percentage points of accuracy.')
```

Both tuned models correctly classify 112 of 114 test cases. The tuned logistic regression misses two malignant cases with no false alarms; SVM misses one malignant case and gives one false alarm. SVM has the highest test recall and F1, while the selected and tuned logistic regressions have the highest AUC, showing that these metrics describe different aspects of performance.

The cross-validation standard deviations show variation across folds, not confidence intervals. Small differences on one test split should be interpreted cautiously.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
for ax, (name, pred) in zip(axes.flat, predictions.items()):
    sns.heatmap(confusion_matrix(y_test, pred, labels=[0,1]), annot=True, fmt='d',
                cmap='Blues', cbar=False, vmin=0, vmax=72, square=True,
                xticklabels=['Benign','Malignant'], yticklabels=['Benign','Malignant'], ax=ax)
    ax.set(title=name, xlabel='Predicted diagnosis', ylabel='Actual diagnosis')
fig.suptitle('Held-out confusion matrices: 72 benign and 42 malignant cases')
plt.tight_layout(); plt.show()
fig, ax = plt.subplots(figsize=(7, 5))
for name, score in scores.items():
    fpr, tpr, _ = roc_curve(y_test, score)
    ax.plot(fpr, tpr, label=f'{name} (AUC = {roc_auc_score(y_test, score):.4f})')
ax.plot([0,1], [0,1], '--', color='gray', label='Chance ranking')
ax.set(title='Held-out ROC curves', xlabel='False positive rate', ylabel='True positive rate'); ax.legend(loc='lower right')
plt.tight_layout(); plt.show()
```

Confusion matrices show the types of errors: false alarms in the upper-right cell and missed malignant cases in the lower-left cell. ROC curves compare how well models separate the classes across different thresholds. A high AUC does not necessarily mean that every malignant case is detected at the chosen threshold.

## 9. Conclusions and limitations
Among the six models compared, logistic regression with all 30 features achieved the highest average cross-validation F1 score (0.9640), making it the preferred model. Reducing the number of features to 15 lowered performance, suggesting that some discarded features provided useful information. Hyperparameter tuning improved the reduced model, and SVM performed slightly better than the tuned logistic regression, but neither exceeded the full-feature baseline in cross-validation.

On the test set, both tuned models achieved 98.25% accuracy, compared with 97.37% for the baseline. However, they made different errors: tuned logistic regression missed two malignant cases with no false alarms, while SVM missed one malignant case and produced one false alarm. This shows why accuracy should be considered alongside recall, precision, and F1. The small test-set advantage does not provide enough evidence to replace the model selected through cross-validation.

These findings are limited to a small dataset from one source. Model selection can produce optimistic cross-validation scores, and the test set was also examined during an earlier version of the analysis, so it does not provide a completely independent final check. Overall, the results show that reducing features or using a more complex model does not necessarily improve classification.
